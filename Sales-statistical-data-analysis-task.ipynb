```python
# mean=df2.mean(axis = 'A')
mean=df2.mean(axis=1)
medians = df2.median(axis=1)
mode = df2.mode(axis=1)

print(f'mean is \n {mean}')
print(f'medians is \n {medians}')
print(f'mode is \n {mode}')
```
